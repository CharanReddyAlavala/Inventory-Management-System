```python
from pyspark.sql.functions import *

inv = spark.read.table("retail_catalog.bronze.inventory_events")
sku = spark.read.table("retail_catalog.bronze.sku_master")
wh  = spark.read.table("retail_catalog.bronze.warehouse_master")
```

```python
# Create silver versions of SKU and warehouse data
silver_sku = sku.drop("_rescued_data") if "_rescued_data" in sku.columns else sku
silver_wh = wh.drop("_rescued_data") if "_rescued_data" in wh.columns else wh
```

```python
from pyspark.sql.functions import *

# -------------------------
# STEP 0: RELOAD FULL DATA
# -------------------------
inv = spark.table("retail_catalog.bronze.inventory_events")

# -------------------------
# STEP 1: Clean base
# -------------------------
silver_inv = (
    inv
    .withColumn("sku_id", trim(col("sku_id")))
    .withColumn("warehouse_id", trim(col("warehouse_id")))
)

# -------------------------
# STEP 2: Fix NULL warehouse_id 
# -------------------------
# Get one valid warehouse_id using aggregation
replacement_wh = (
    silver_inv
    .filter(col("warehouse_id").isNotNull())
    .agg(first("warehouse_id").alias("wh"))
    .first()["wh"]
)

silver_inv = silver_inv.withColumn(
    "warehouse_id",
    when(col("warehouse_id").isNull(), lit(replacement_wh))
    .otherwise(col("warehouse_id"))
)

# -------------------------
# STEP 3: Safe date parsing
# -------------------------
silver_inv = silver_inv.withColumn(
    "parsed_time",
    expr("""
        coalesce(
            try_to_timestamp(event_time, 'yyyy/MM/dd HH:mm'),
            try_to_timestamp(event_time, 'yyyy-MM-dd HH:mm:ss'),
            try_to_timestamp(event_time, 'yyyy-MM-dd'),
            try_to_timestamp(event_time, 'dd/MM/yyyy')
        )
    """)
)

# -------------------------
# STEP 4: Replace NULL dates
# -------------------------
median_date = (
    silver_inv
    .filter(col("parsed_time").isNotNull())
    .agg(expr("percentile_approx(parsed_time, 0.5)").alias("median_date"))
    .first()["median_date"]
)

silver_inv = silver_inv.withColumn(
    "event_time",
    when(col("parsed_time").isNull(), lit(median_date))
    .otherwise(col("parsed_time"))
).drop("parsed_time")

silver_inv = silver_inv.withColumn(
    "event_date",
    to_date(col("event_time"))
)

# -------------------------
# STEP 5: Net movement
# -------------------------
silver_inv = silver_inv.withColumn(
    "net_qty",
    when(col("movement_type") == "IN", col("quantity"))
    .otherwise(-col("quantity"))
)

silver_inv = silver_inv.withColumn(
    "movement_type",
    when(upper(col("movement_type")).isin("IN", "PURCHASE"), "IN")
    .when(upper(col("movement_type")).isin("OUT", "SALE"), "OUT")
    .otherwise("OUT")   # all invalid → OUT
)

# -------------------------
# STEP 6: Replace negative values
# -------------------------
avg_qty = (
    silver_inv
    .filter(col("net_qty") > 0)
    .agg(round(avg("net_qty"), 0).alias("avg_qty"))
    .first()["avg_qty"]
)

silver_inv = silver_inv.withColumn(
    "net_qty",
    when(col("net_qty") < 0, lit(avg_qty))
    .otherwise(col("net_qty"))
)

# -------------------------
# STEP 7: Cleanup
# -------------------------
silver_inv = silver_inv.withColumn(
    "processed_timestamp",
    current_timestamp()
)

if "_rescued_data" in silver_inv.columns:
    silver_inv = silver_inv.drop("_rescued_data")

display(silver_inv)
```

```python
from pyspark.sql.functions import *

# -------------------------
# STEP 1: Strong normalization (CRITICAL FIX)
# -------------------------
def clean_id(c):
    return regexp_replace(upper(trim(col(c))), "[^A-Z0-9]", "")

silver_inv_clean = (
    silver_inv
    .withColumn("sku_id", clean_id("sku_id"))
    .withColumn("warehouse_id", clean_id("warehouse_id"))
    .drop("ingestion_timestamp")
)

silver_sku_clean = (
    silver_sku
    .withColumn("sku_id", clean_id("sku_id"))
    .drop("ingestion_timestamp")
)

silver_wh_clean = (
    silver_wh
    .withColumn("warehouse_id", clean_id("warehouse_id"))
    .drop("ingestion_timestamp")
)

# -------------------------
# STEP 2: Join (NO DATA LOSS)
# -------------------------
silver_final = (
    silver_inv_clean
    .join(silver_sku_clean, "sku_id", "left")
    .join(silver_wh_clean, "warehouse_id", "left")
)

# -------------------------
# STEP 3: Handle NULLs safely
# -------------------------
silver_final = (
    silver_final
    .withColumn("category", coalesce(col("category"), lit("UNKNOWN")))
    .withColumn("location", coalesce(col("location"), lit("UNKNOWN")))
    .withColumn("unit_cost", coalesce(col("unit_cost"), lit(0.0)))
    .withColumn("reorder_point", coalesce(col("reorder_point"), lit(50)))
    .withColumn("capacity", coalesce(col("capacity"), lit(0)))
)

silver_final = silver_final.filter(
    (col("category") != "UNKNOWN") &
    (col("location") != "UNKNOWN")
)

# -------------------------
# STEP 4: Save final table
# -------------------------
silver_final.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("retail_catalog.silver.inventory_conformed")

display(silver_final)
```
